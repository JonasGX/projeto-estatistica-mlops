## Percentuais

```python
dados = {'emprego': ['Adm_banco_de_dados', 'Programador', 'Arquiteto_de_Redes'],
         'nova_jersey': [97350, 82080, 112840],
         'florida': [77140, 71540, 62310]
         }

dados
```

```python
dataset = pd.DataFrame(dados)
dataset
```

```python
dataset['nova_jersey'].sum()
```

```python
dataset['florida'].sum()
```

```python
dataset['nova_jersey_%'] = (dataset['nova_jersey'] / dataset['nova_jersey'].sum()) * 100
dataset['florida_%'] = (dataset['florida'] / dataset['florida'].sum()) * 100

```

## Exercicios Percentuais

```python
census = pd.read_csv('../Bases de dados/census.csv')
```

```python
census = census[['education', 'income']]
```

```python
rendas = census.groupby(['education', 'income'])['education'].count()

rendas
```

```python
rendas.index
```

```python
rendas[' Bachelors', ' <=50K'], rendas[' Bachelors', ' >50K']
```

```python
total = int(rendas[' Bachelors', ' >50K']) + int(rendas[' Bachelors', ' <=50K'])
```

```python
dados = {'salario': ['>50K', '<=50K'],
         '_%_': [(rendas[' Bachelors', ' >50K'] / total) * 100, (rendas[' Bachelors', ' <=50K'] / total) * 100]
         }

result = pd.DataFrame(dados)
result
```
