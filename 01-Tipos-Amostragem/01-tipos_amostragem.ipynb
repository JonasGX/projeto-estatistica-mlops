## Amostragem aleatória simples

```python
# sample -> seleciona a quantidade de amostras de uma população
# random_state=N -> Cria um identificador para a sua amostra.
df_amostra_simples = df_census.sample(n = 100, random_state=1)

df_amostra_simples.shape
```

```python
df_amostra_simples.head()
```

```python
# random_state -> Os resultados de amostragem serão os mesmos em outra máquina. Isso ocorre pois a base de dados é idêntica  possuindo os mesmos dados e padrão de escolha.

def amostragem_aleatoria_simples(dataset: pd.DataFrame,
                              qtd_amostras: int,
                              id_amostra: int) -> pd.DataFrame:
    df = dataset.sample(n=qtd_amostras, random_state=id_amostra)

    return df
```

```python
df_amostra_simples = amostragem_aleatoria_simples(df_census, 100, 1)
```

```python
df_amostra_simples.head()
```

## Amostragem sistemática

```python
def amostragem_sistematica(qtd_amostra, df):
    calculo = math.trunc(len(df) // qtd_amostra)
    random.seed(1)
    valor_inicial = random.randint(0, calculo)
    indices = np.arange(valor_inicial, len(df), step=calculo)

    amostra = df.iloc[indices]
    return amostra

```

```python
df_amostra_sistematica = amostragem_sistematica(100, df_census)
```

## Amostragem por grupos

```python
def amostragem_agrupamento(df, qtd_grupos):
    # df_copy = df.copy()
    qtd_registros = len(df) // qtd_grupos
    grupos = []
    id_grupo = 1
    contagem = 0
    # Itera sobre todas as linhas do df
    for _ in df.iterrows():
        grupos.append(id_grupo)
        contagem += 1
        if contagem > qtd_registros:
            contagem = 0
            id_grupo += 1

    # Cria a coluna grupos
    df['grupo'] = grupos

    random.seed(1)
    grupo_sort = random.randint(1,qtd_grupos)
    return df[df['grupo'] == grupo_sort]
```

```python
df_amostra_grupo = amostragem_agrupamento(df_census, 325)
```

```python
# Queremos 100 registros como amostra. Precisamos identificar esse valor em porcentagem
porcent = 100 / len(df_census)
```

```python
porcent
```

```python
# Biblioteca que faz o calculo da amostra estratificada
from sklearn.model_selection import StratifiedShuffleSplit

def amostragem_estratificada(df, porcent):
    # StratifiedShuffleSplit -> Utilizada para buscar a quantidade de registros de uma base. É muito utilizado para treinar modelos de MLops
    split = StratifiedShuffleSplit(test_size=porcent)
    # Primeiro argumento é a base de dados de treinamento e o segundo argumento é a base de dados de teste
    # Se escolhermos test_size = 20%, primeiro argumento terá 80% e o segundo terá 20%
    for _, y in split.split(df, df['income']):
        df_y = df.iloc[y]

    return df_y
```

```python
df_estratificada = amostragem_estratificada(df_census, porcent)
```

## Amostragem de reservatório

```python
def amostragem_reservatorio(dataset, amostras):
    tamanho = len(dataset)
    stream = []

    for i in range(tamanho):
        stream.append(i)

    i = 0
    reservatorio = [0] * amostras

    for i in range(amostras):
        reservatorio[i] = stream[i]


    while i < tamanho:
        j = random.randrange(i + 1)
        if j < amostras:
            reservatorio[j] = stream[i]
        i += 1

    return dataset.iloc[reservatorio]
```

```python
df_reservatorio = amostragem_reservatorio(df_census, 100)
```
